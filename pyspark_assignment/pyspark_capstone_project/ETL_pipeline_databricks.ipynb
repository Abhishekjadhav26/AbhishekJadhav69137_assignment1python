```python
# verify tables code
bronze_customers = spark.read.table(
    "retail_demo.bronze_customers"
)

bronze_products = spark.read.table(
    "retail_demo.bronze_products"
)

bronze_sales = spark.read.table(
    "retail_demo.bronze_sales"
)

print("  BRONZE CUSTOMERS  ")
display(bronze_customers)

print(" BRONZE PRODUCTS  ")
display(bronze_products)

print("  BRONZE SALES  ")
display(bronze_sales)
```

```python
print("  CUSTOMER NULL VALUES  ")

for c in bronze_customers.columns:
    count_null = bronze_customers.filter(
        col(c).isNull()
    ).count()

    print(c, count_null)
```

```python
print(" PRODUCT NULL VALUES  ")

for c in bronze_products.columns:
    count_null = bronze_products.filter(
        col(c).isNull()
    ).count()

    print(c, count_null)
```

```python
print("  SALES NULL VALUES  ")

for c in bronze_sales.columns:
    count_null = bronze_sales.filter(
        col(c).isNull()
    ).count()

    print(c, count_null)
```

```python
bronze_customers.groupBy(
    "customer_id"
).count().filter(
    "count > 1"
).show()
```

```python
bronze_products.groupBy(
    "product_id"
).count().filter(
    "count > 1"
).show()
```

```python
bronze_sales.groupBy(
    "transaction_id"
).count().filter(
    "count > 1"
).show()
```

```python
silver_customers = bronze_customers \
    .dropDuplicates(["customer_id"]) \
    .dropna(subset=[
        "customer_id",
        "customer_name"
    ]) \
    .withColumn(
        "customer_name",
        trim(col("customer_name"))
    ) \
    .withColumn(
        "gender",
        trim(col("gender"))
    ) \
    .withColumn(
        "city",
        trim(col("city"))
    )

display(silver_customers)
```

```python
silver_products = bronze_products \
    .dropDuplicates(["product_id"]) \
    .dropna(subset=[
        "product_id",
        "product_name",
        "category",
        "unit_price"
    ]) \
    .filter(
        col("unit_price") > 0
    ) \
    .withColumn(
        "product_name",
        trim(col("product_name"))
    ) \
    .withColumn(
        "category",
        trim(col("category"))
    ) \
    .withColumn(
        "brand",
        trim(col("brand"))
    )

display(silver_products)
```

```python
silver_sales = bronze_sales \
    .dropDuplicates(["transaction_id"]) \
    .dropna(subset=[
        "transaction_id",
        "customer_id",
        "product_id",
        "quantity_sold",
        "transaction_date"
    ]) \
    .filter(
        col("quantity_sold") > 0
    ) \
    .withColumn(
        "transaction_date",
        to_date("transaction_date")
    )

display(silver_sales)
```

```python
# join sales with customers
silver_enriched = silver_sales.join(
    silver_customers,
    "customer_id",
    "inner"
)
```

```python
#join with products
silver_enriched = silver_enriched.join(
    silver_products,
    "product_id",
    "inner"
)
```

```python
silver_enriched = silver_enriched \
    .withColumn(
        "revenue",
        col("quantity_sold") * col("unit_price")
    ) \
    .withColumn(
        "transaction_year",
        year(col("transaction_date"))
    ) \
    .withColumn(
        "transaction_month",
        month(col("transaction_date"))
    )

display(silver_enriched)
```

```python
silver_customers.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "retail_demo.silver_customers"
    )

silver_products.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "retail_demo.silver_products"
    )

silver_enriched.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "retail_demo.silver_retail"
    )
```

```python
gold_category = silver_enriched.groupBy(
    "category"
).agg(
    sum("revenue").alias("total_revenue")
).orderBy(
    desc("total_revenue")
)
display(gold_category)
```

```python
gold_city = silver_enriched.groupBy(
    "city"
).agg(
    sum("revenue").alias("total_revenue")
).orderBy(
    desc("total_revenue")
)
display(gold_city)
```

```python
gold_customer = silver_enriched.groupBy(
    "customer_id",
    "customer_name"
).agg(
    sum("revenue").alias("customer_revenue")
).orderBy(
    desc("customer_revenue")
)
display(gold_customer)
```

```python
gold_monthly = silver_enriched.groupBy(
    date_format(
        "transaction_date",
        "yyyy-MM"
    ).alias("month")
).agg(
    sum("revenue").alias("monthly_revenue")
).orderBy(
    "month"
)
display(gold_monthly)
```

```python
gold_products = silver_enriched.groupBy(
    "product_id",
    "product_name"
).agg(
    sum("revenue").alias("product_revenue")
).orderBy(
    desc("product_revenue")
)
display(gold_products)
```

```python
product_revenue_category = gold_products.withColumn(
    "revenue_category",
    when(
        col("product_revenue") >= 50000,
        "High Contribution"
    ).when(
        col("product_revenue") >= 20000,
        "Medium Contribution"
    ).otherwise(
        "Low Contribution"
    )
)
display(product_revenue_category)
```

```python
customer_frequency = silver_enriched.groupBy(
    "customer_id",
    "customer_name"
).agg(
    count("transaction_id").alias("purchase_frequency")
).orderBy(
    desc("purchase_frequency")
)
display(customer_frequency)
```

```python
high_value_customers = gold_customer.withColumn(
    "customer_category",
    when(
        col("customer_revenue") >= 50000,
        "High Value"
    ).otherwise("Regular")
)
display(high_value_customers)
```

```python
#Total Revenue calculation
silver_enriched.agg(
    sum("revenue").alias("total_revenue")
).show()

#Total Orders calculation
silver_enriched.agg(
    count("transaction_id").alias("total_orders")
).show()
```

```python
silver_enriched.agg(
    round(
        avg("revenue"),
        2
    ).alias("average_order_value")
).show()
```

```python
#Total customers
silver_enriched.select(
    "customer_id"
).distinct().agg(
    count("customer_id").alias("total_customers")
).show()
```

```python
active_customers = silver_enriched.select(
    "customer_id"
).distinct()

print("Active Customers:", active_customers.count())
```

```python
#repeat customers
repeat_customers = silver_enriched.groupBy(
    "customer_id",
    "customer_name"
).agg(
    count("transaction_id").alias("order_count")
).filter(
    col("order_count") > 1
)
display(repeat_customers)
```

```python
top_10_products = gold_products.limit(10)
display(top_10_products)
```

```python
top_10_customers = gold_customer.limit(10)
display(top_10_customers)
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import lag

month_window = Window.orderBy("month")

gold_monthly_growth = gold_monthly.withColumn(
    "previous_month_revenue",
    lag("monthly_revenue").over(month_window)
).withColumn(
    "monthly_growth_percent",
    round(
        (
            (col("monthly_revenue") - col("previous_month_revenue"))
            / col("previous_month_revenue")
        ) * 100,
        2
    )
)
display(gold_monthly_growth)
```

```python
from pyspark.sql.functions import row_number

product_window = Window.orderBy(
    desc("product_revenue")
)
product_ranking = gold_products.withColumn(
    "rank",
    row_number().over(product_window)
)
display(product_ranking)
```

```python
category_performance = silver_enriched.groupBy(
    "category"
).agg(
    sum("revenue").alias("total_revenue"),
    sum("quantity_sold").alias("total_quantity"),
    count("transaction_id").alias("total_orders")
).orderBy(
    desc("total_revenue")
)
display(category_performance)
```

```python
best_selling_products = silver_enriched.groupBy(
    "product_id",
    "product_name"
).agg(
    sum("quantity_sold").alias("total_quantity_sold")
).orderBy(
    desc("total_quantity_sold")
)
display(best_selling_products)
```
